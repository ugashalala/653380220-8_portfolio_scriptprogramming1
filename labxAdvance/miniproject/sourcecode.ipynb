```python
import random


# ---------- เก็บข้อมูลสกิลที่ผู้เล่นสามารถเลือกได้ทั้งหมด -----------

skills = {
    "1": {"name": "attack", "type": "damage", "value": (10, 20)},
    "2": {"name": "punchofluck", "type": "damage", "value": (0, 50)},
    "3": {"name": "heal", "type": "heal", "value": 25},
    "4": {"name": "defend", "type": "defend"}
}
```

```python
# ---------- func หน้าเมนูของ player ----------

def get_player_action():
    while True:
        print("\nChoose action:")
        print("[1] Attack ⚔️")
        print("[2] Punch of Luck 🍀")
        print("[3] Heal 💊")
        print("[4] Defend 🛡️")

        choice = input("Enter number (1-4): ").strip()

# ---------- input validation ----------

        if choice in skills: 
            return choice #ถ้าข้อมูลที่ป้อนเข้ามามีอยู้ใน skill ก็จะส่งค่านั้นออกไปเลยแล้วจบ while loop
        else:
            print("❌ Invalid input! Try again.") #ถ้าข้อมูลที่ป้อนเข้ามาไม่มีอยู้ใน skill ก็จะ print "invalid input" แล้ววน while loop เพื่อให้ player ป้อนข้อมูลใหม่
```

```python
def main():

    # set ค่า hp ของ player กับ enemy
    player_hp = 100
    enemy_hp = 100
    player_status = {"defend": False}

    #set ค่า turn เริ่มต้นเพื่อให้ทุกครั้งที่ run จะเริ่มที่ turn 1 เสมอ
    turn = 1  

    print("⚔️ === Turn-Based Battle Game === ⚔️") #print title game before loop start

    #เข้า loop การเล่นเกมโดยเกมจะจบจนกว่า hp ของ player หรือ enemy จะเหลือ 0
    while player_hp > 0 and enemy_hp > 0:
        print(f"\n--- Turn {turn} ---") #แสดงว่าตอนนี้ถึง turn ที่เท่าไหร่
        print(f"❤️ Player HP: {player_hp} | 👾 Enemy HP: {enemy_hp}") #แสดงค่า hp ของ player และ enemy

        # ===== Player Turn =====
        choice = get_player_action() #เก็บค่าที่ส่งมาจาก func get_player_action() ไว้ใน choice
        skill = skills[choice] #ดึง value จาก key ที่ตรงกับ choice

        print(f"\n👉 You selected: {skill['name'].upper()}") #แสดง action ที่เลือก

        #ถ้า player เลือก action ที่มี type damage
        if skill["type"] == "damage":
            damage = random.randint(*skill["value"]) #random จาก range ใน dictionary ใน skill
            enemy_hp -= damage #ลบเลือดของ enemy ตามดาเมจที่ random ค่าออกมาได้
            print(f"\n⚔️ You used {skill['name']} and deal {damage} damage!") #แสดงชื่อสกิลที่เลือกและดาเมจที่ทำได้

        #ถ้า player เลือก action ที่มี type heal
        elif skill["type"] == "heal":
            player_hp = min(player_hp + skill["value"], 100) #เพิ่มเลือด player โดยกำหนดให้ไม่เกิน 100
            print(f"\n💊 You heal for {skill['value']} HP! And now You have {player_hp} HP!") #แสดง hp ของผู้เล่นว่า heal ไปเท่าไหร่ และ hp ตอนนี้เหลือเท่าไหร่

        #ถ้า player เลือก action ที่มี type defend
        elif skill["type"] == "defend":
            #ถ้า player เลือก defend จะเปลี่ยนค่า status ให้เป็น true เพื่อ set ว่าเทิร์นนั้น player เลือก player และจะได้นำไปใช้ในส่วนของ enemy attack 
            player_status["defend"] = True
            print("\n🛡️ You are defending this turn!")

        # ===== Check Enemy Dead =====
        if enemy_hp <= 0:
            print("\n🎉 You Win!")
            break #จบการทำงานเมื่อ enemy hp = 0

        # ===== Enemy Turn =====
        if enemy_hp < 30: #เช็คว่าถ้า enemy hp < 30 จะให้ enemy heal โดยอัตโนมัติ
            enemy_action = "heal"
        else:
            enemy_action = random.choice(["attack", "attack", "heal"]) #ให้ enemy สุ่ม action ใน list โดย ใน list มี attack 2 ครั้งเพื่อเพิ่มโอกาสในการโจมตีของ enemy

        if enemy_action == "attack":
            damage = random.randint(10, 30) #สุ่มดาเมจในการโจมตีของ enemy ด้วยการ random int 10-30

            if player_status["defend"]:
                damage //= 2 #ถ้า player เลือก action defend ดาเมจที่ enemy สุ่มได้จะถูกเอามา / 2
                player_status["defend"] = False #หลังจาก / 2 ดาเมจแล้ว จะเซ็ต status กลับไปที่ false
                print(f"🛡️ You reduced the damage! from {damage*2} to {damage} ")

            player_hp -= damage #ลบเลือดของ player ตามดาเมจที่ enemy ทำได้
            print(f"👾 Enemy attacks and deals {damage} damage!")

        #ในกรณีถ้า action ของ enemy ไม่ใช่ attack
        else:
            heal = 15
            enemy_hp = min(enemy_hp + heal, 100)
            print(f"👾 Enemy heals for {heal} HP!")

        # ===== Check Player Dead =====
        if player_hp <= 0:
            print("\n💀 You Lose!")
            break #จบการทำงานเมื่อ player hp = 0

        turn += 1 #เพิ่มเลข turn ที่จะแสดงถ้ายังไม่จบ loop

    print("\n=== END ===")
```

# Run Program

```python
main()
```
