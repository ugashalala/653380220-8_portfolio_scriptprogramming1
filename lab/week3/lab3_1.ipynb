```python
# Lab 3.1: Multiplication Table Generator

# Part 1: Single Number Table

num = int(input("Enter a number: "))

print(f"\nMultiplication Table of {num}")

for i in range(1, 13):
    print(f"{num} x {i} = {num * i}")

```

```python
# Part 2: Full Multiplication Grid

print("\nFull Multiplication Grid (12x12)")

for i in range(1, 13):
    for j in range(1, 13):
        print(f"{i*j:4d}", end="")
    print()
```
