```python
# Activity 3.2: Pattern Printing with Nested Loops

# Pattern 1: Square of Asterisks

print("Pattern 1: Square")

for i in range(5):
    for j in range(5):
        print("*", end="")
    print()


# Pattern 2: Right-Angled Triangle

print("\nPattern 2: Right-Angled Triangle")

for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()


# Pattern 3: Inverted Right-Angled Triangle

print("\nPattern 3: Inverted Triangle")

for i in range(5, 0, -1):
    for j in range(i):
        print("*", end="")
    print()


# Pattern 4: Pyramid (Challenge)

print("\nPattern 4: Pyramid")

for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()
```
