```python
# Puzzler 1
x = 10
y = 3.5
z = x + y
print(z)
print(type(z))
```

ถ้าเอา int มาบวกกับ float ตัวภาษาจะเปลี่ยนให้ int เป็น float โดยอัตโนมัติ

```python
# Puzzler 2
name = "Alice"
age = "30" # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))
```
