# Activity 1.1: Class Discussion - Uses of Scripting
* Format: Open discussion, possibly in small groups first, then full class.
* Prompt: "Based on the introduction, where do you think scripting with Python could be applied in real-world scenarios? Think about tasks you do daily or things you've seen in other courses."
* Expected outcomes: Students will brainstorm and share ideas like automating file operations, simple data analysis, web development, game scripting, scientific computing, etc. Instructor will guide the discussion to highlight Python's versatility.


คำตอบ:
* ลดงานซ้ำซ้อน (Automation): ใช้เปลี่ยนชื่อไฟล์ทีละเยอะ ๆ, จัดหมวดหมู่เอกสารอัตโนมัติ, หรือสั่งให้พิมพ์/กรอกข้อมูลใน Excel และ PDF แทนเรา

* จัดการข้อมูล (Data & Web Scraping): เขียนสคริปต์ดึงข้อมูลจากหน้าเว็บ (เช่น ราคาสินค้า, สภาพอากาศ) แล้วนำมาคำนวณ ทำสถิติ หรือพลอตกราฟสรุปผล
