# Lab 2.2: Simple Decision-Making Program (Movie Rating)
**Objective:** Students will create a program that recommends a movie rating based on the user's age, demonstrating nested or chained conditionals.

**Tasks (Common for Local & Colab):**
1. Create a new Python file (lab2_2.py) or Colab code cell.
2. Prompt the user to enter their age. Convert it to an integer.
3. Implement the following logic:
* If age is less than 5: Print "You're too young for movies! Enjoy cartoons."
* If age is 5 to 12 (inclusive): Recommend "G-rated or PG-rated movies."
* If age is 13 to 17 (inclusive): Recommend "PG-13 or R-rated (with parental guidance)."
* If age is 18 or older: Recommend "Any movie rating."
4. **Challenge:** Add a secondary prompt: "Do you like action movies? (yes/no)". If the age is 18 or older AND they like action, print "You might enjoy the latest action blockbuster!" (This requires a nested if or more complex logical and conditions).


```python
# รับค่าอินพุตจากผู้ใช้
age_input = input("กรุณากรอกอายุของคุณ (ตัวเลขจำนวนเต็ม): ")
print("อายุ : ", age_input)

# ตรวจสอบว่าผู้ใช้ป้อนตัวเลขจริงและไม่ติดลบ
if not age_input.isdigit():
    print("❌ ข้อผิดพลาด: กรุณากรอกอายุเป็นตัวเลขจำนวนเต็มบวกเท่านั้น")
    print("💡 [คำอธิบายของระบบ]: ระบบปฏิเสธการคำนวณเพื่อป้องกันข้อผิดพลาดทางตรรกะ")
else:
    age = int(age_input)

# ดักจับค่าอายุที่เกินจริง
if age > 100:
    print("⚠️ warning: อายุที่ระบุสูงเกินความเป็นจริง โปรดตรวจสอบอีกครั้ง")

elif age < 5:
    recommendation = "You're too young for movies! Enjoy cartoons."
    print(f"🎬 คำแนะนำสำหรับคุณ: {recommendation}")

elif 5 <= age <= 12:
    recommendation = "G-rated or PG-rated movies."
    print(f"🎬 คำแนะนำสำหรับคุณ: {recommendation}")

elif 13 <= age <= 17:
    recommendation = "PG-13 or R-rated (with parental guidance)."
    print(f"🎬 คำแนะนำสำหรับคุณ: {recommendation}")

else:
    recommendation = "Any movie rating."
    print(f"🎬 คำแนะนำสำหรับคุณ: {recommendation}")
    # รับข้อมูลความชอบ และใช้ .strip().lower() เพื่อให้ระบบยืดหยุ่นต่อพิมพ์เล็ก/ใหญ่ หรือเคาะเว้นวรรค
    like_action = input("\n💡 Do you like action movies? (yes/no)").strip().lower()
    if like_action == "yes":
        print("💥 You might enjoy the latest action blockbuster!")



  
```
