```python
# Lab 4.2: Geometric Calculations
# Tasks 2-5

import math

# 2. Define Points
point1 = (3, 4)
point2 = (6, 8)

# 3. Access Coordinates
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# 4. Attempt Modification (Immutability)

# Uncomment the next line to observe the TypeError
# point1[0] = 5

# 5. Calculate Distance
x1, y1 = point1
x2, y2 = point2

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print(f"Distance between points: {distance}")
```

```python
# Lab 4.2: Challenge
# Calculate Distance Using User Input

import math

# Enter coordinates for Point 1
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

# Enter coordinates for Point 2
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

# Store coordinates in tuples
point1 = (x1, y1)
point2 = (x2, y2)

# Calculate distance
distance = math.sqrt((point2[0] - point1[0])**2
                     + (point2[1] - point1[1])**2)

print("Point 1:", point1)
print("Point 2:", point2)
print(f"Distance between points: {distance}")
```
