```python
# Lab 4.1: Student Management System
# Tasks 2-7

# 2. Initialize an empty list
student_names = []

# 3. Add Students
for i in range(5):
    name = input(f"Enter student name {i + 1}: ")
    student_names.append(name)

print("\nStudent List:", student_names)

# 4. Find a Student
search_name = input("\nEnter a name to search: ")

if search_name in student_names:
    print("Student found!")
    print("Index:", student_names.index(search_name))
else:
    print("Student not found.")

# 5. Remove a Student
remove_name = input("\nEnter a name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name)
    print("Student removed successfully!")
else:
    print("Student not found.")

print("Updated Student List:", student_names)

# 6. Sort Students
student_names.sort()
print("Sorted Student List:", student_names)

# 7. Count Students
print("Total students:", len(student_names))
```

```python
# Lab 4.1: Challenge
# Menu-Driven Student Management System

while True:
    print("\n=== Student Management System ===")
    print("1. Add Student")
    print("2. Remove Student")
    print("3. Search Student")
    print("4. Sort Students")
    print("5. Count Students")
    print("6. Display All Students")
    print("7. Exit")

    choice = input("Choose an option (1-7): ")

    if choice == "1":
        name = input("Enter student name to add: ")
        student_names.append(name)
        print("Student added successfully!")

    elif choice == "2":
        name = input("Enter student name to remove: ")

        if name in student_names:
            student_names.remove(name)
            print("Student removed successfully!")
        else:
            print("Student not found.")

    elif choice == "3":
        name = input("Enter student name to search: ")

        if name in student_names:
            print("Student found!")
            print("Index:", student_names.index(name))
        else:
            print("Student not found.")

    elif choice == "4":
        student_names.sort()
        print("Sorted Student List:", student_names)

    elif choice == "5":
        print("Total students:", len(student_names))

    elif choice == "6":
        print("Student List:", student_names)

    elif choice == "7":
        print("Exiting program. Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
