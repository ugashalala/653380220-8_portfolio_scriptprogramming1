Instructions: Present students with a list of real-world data scenarios. For each
scenario, they must decide whether a list or a tuple would be the more appropriate
Python data structure and explain their reasoning (focus on mutability/immutability and
ordered nature).
○ Scenarios:
1. A collection of daily temperatures recorded over a week.

- list	เพราะสามารถแก้ไขอุณหภูมิของแต่ละวันได้ หากพบว่าบันทึกข้อมูลผิด

2. The RGB color values for a specific shade (e.g., (255, 0, 0) for red).

- tuple เพราะค่าสี RGB ทั้งสามค่ามักถูกใช้เป็นชุดข้อมูลเดียวกันและไม่ควรเปลี่ยนแปลงโดยไม่ตั้งใจ
3. Items in a shopping cart that can be added or removed.

- list	เพราะต้องเพิ่มสินค้าใหม่หรือลบสินค้าออกได้

4. The days of the week in order.

- tuple เพราะเป็นชุดข้อมูลที่มีลำดับแน่นอนและโดยทั่วไปไม่จำเป็นต้องแก้ไข

5. Student scores on a series of assignments, which might be updated.

- list	เพราะสามารถแก้ไขคะแนนเดิมหรือเพิ่มคะแนนของงานใหม่ได้

6. GPS coordinates for a fixed landmark.

- tuple เพราะพิกัด (latitude, longitude)เป็นข้อมูลคู่ที่มีความสัมพันธ์กันและโดยทั่วไปไม่ควรเปลี่ยนแปลง


# Example

```python
# 1. Daily temperatures
temperatures = [30, 32, 31, 29, 33, 34, 30]

# 2. RGB color values
red = (255, 0, 0)

# 3. Shopping cart
shopping_cart = ["Milk", "Bread", "Eggs"]
shopping_cart.append("Butter")
shopping_cart.remove("Bread")

# 4. Days of the week
days = ("Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday")

# 5. Student scores
scores = [80, 75, 90]
scores[1] = 85

# 6. GPS coordinates
landmark = (16.4419, 102.8350)

print("Temperatures:", temperatures)
print("Red RGB:", red)
print("Shopping cart:", shopping_cart)
print("Days:", days)
print("Student scores:", scores)
print("GPS coordinates:", landmark)
```
