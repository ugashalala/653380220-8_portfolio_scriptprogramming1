```python
# Activity 6.1: Refactoring a Previous Lab Assignment

# Function to calculate the multiplication result
def multiply(a, b):
    return a * b
```

```python
# Function to display a multiplication table for one number
def display_table(number):
    print(f"\nMultiplication Table of {number}")

    for i in range(1, 13):
        result = multiply(number, i)
        print(f"{number} x {i} = {result}")
```

```python
# Main program
number = int(input("Enter a number (1-12): "))

if 1 <= number <= 12:
    display_table(number)
else:
    print("Please enter a number between 1 and 12.")
```
