# Part 1: Benefits of Modular Programming
## Question: What are the benefits of modular programming?

Modular programming has several benefits:
1. Code Reusability: Functions and modules can be reused in different parts of a program without rewriting the same code.
2. Readability: Breaking a large program into smaller functions makes the code easier to understand.
3. Teamwork: Different team members can work on separate modules at the same time.
4. Debugging: Errors are easier to identify because each function has a specific responsibility.
5. Testing: Individual functions can be tested separately to ensure they work correctly.
6. Maintainability: Updating one module is easier and less likely to affect unrelated parts of the program.
Example: In a student management system, we can create separate functions for adding students, searching for students, and calculating grades. This makes the program easier to develop, test, and maintain.

# Part 2: Standard Library Exploration
## Official documentation: Python Standard Library

เลือกศึกษา 3 โมดูลต่อไปนี้

1. datetime — Date and Time

- ใช้จัดการวันที่ เวลา และคำนวณระยะเวลาระหว่างวันที่

```python
import datetime

today = datetime.date.today()
print("Today:", today)
```

ตัวอย่างการใช้งาน: แสดงวันที่ปัจจุบันหรือบันทึกวันที่สมัครสมาชิก

2. os — Operating System Interaction

- ใช้ทำงานกับระบบปฏิบัติการ เช่น ตรวจสอบโฟลเดอร์และจัดการเส้นทางไฟล์

```python
import os

print("Current directory:", os.getcwd())
```

ตัวอย่างการใช้งาน: ตรวจสอบว่าโปรแกรมกำลังทำงานอยู่ในโฟลเดอร์ใด

3. collections — Specialized Data Structures

- มีโครงสร้างข้อมูลเพิ่มเติม เช่น Counter สำหรับนับจำนวนสิ่งที่ซ้ำกัน

```python
from collections import Counter

fruits = ["apple", "banana", "apple", "orange", "apple"]
print(Counter(fruits))
```

ตัวอย่างการใช้งาน: นับความถี่ของคำในข้อความหรือจำนวนสินค้าที่แต่ละประเภทขายได้
