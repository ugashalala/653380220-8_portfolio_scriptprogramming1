```python
# 1. Dictionary in a game
player_inventory = {
    "sword": 1,
    "potion": 5,
    "shield": 1
}
print("Player inventory:", player_inventory)

# 2. Dictionary in an online store
products = {
    "P001": {"name": "Mouse", "price": 250},
    "P002": {"name": "Keyboard", "price": 590}
}
print("Product P001:", products["P001"])

# 3. Set for user roles
user_roles = {"admin", "editor", "viewer", "admin"}
print("Unique roles:", user_roles)

# 4. Set for blocked IP addresses
blocked_ips = {"192.168.1.10", "10.0.0.5"}

ip_address = "192.168.1.10"

if ip_address in blocked_ips:
    print("Access denied.")
else:
    print("Access allowed.")
```
