```python
# Define student sets
python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

# 1. Students in both courses
both_courses = python_students.intersection(java_students)
print("Students in both courses:", both_courses)

# 2. All unique students
all_students = python_students.union(java_students)
print("All unique students:", all_students)

# 3. Students only in Python
only_python = python_students.difference(java_students)
print("Students only in Python:", only_python)

# 4. Students in only one course
exclusive_students = python_students.symmetric_difference(java_students)
print("Students in only one course:", exclusive_students)

# 5. Remove duplicates from a list
all_students_list = ["S001", "S003", "S003", "S005", "S001", "S002", "S005"]
unique_students_list = list(set(all_students_list))

print("Original list:", all_students_list)
print("List without duplicates:", unique_students_list)
```
