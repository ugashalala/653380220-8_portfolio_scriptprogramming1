```python
# Step 1: Get a sentence from the user
sentence = input("Enter a sentence: ")

# Step 2: Convert the sentence to lowercase
sentence = sentence.lower()

# Step 3: Split the sentence into words
words = sentence.split()

# Step 4: Initialize an empty dictionary
word_counts = {}

# Step 5: Loop through each word
for word in words:

    # Challenge
    word = word.strip('.,!?"')

    # Challenge
    if word == "":
        continue

    # Step 6: Count the frequency of each word
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Step 7: Print the word frequency dictionary
print("Word counts:", word_counts)
```
