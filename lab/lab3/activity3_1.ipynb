```python
# Buggy Snippet A: Infinite Loop
# counter = 0
# while counter < 5:
#     print(counter)
#     # Missing counter increment!
```

Identify the Bug: ไม่มีคำสั่งเพิ่มค่าตัวแปร counter ภายในลูป

Explain Why: เนื่องจากค่า counter เริ่มต้นเป็น 0 และไม่มีการเปลี่ยนแปลงค่า ทำให้เงื่อนไข counter < 5 เป็นจริงตลอดเวลา ส่งผลให้เกิด Infinite Loop

```python
# Snippet A: Fixed Infinite Loop

counter = 0

while counter < 5:
    print(counter)
    counter += 1 #fixed
```

```python
# Buggy Snippet B: Incorrect Range
# for i in range(1, 5): # Goal: print 1, 2, 3, 4, 5
#     print(i)
```

โจทย์ต้องการให้แสดงตัวเลขตั้งแต่ 1 ถึง 5 แต่โค้ดเดิมแสดงเพียง 1 ถึง 4

Identify the Bug: กำหนดค่าสิ้นสุดของ range() ไม่ถูกต้อง

Explain Why: ใน Python ค่าสุดท้ายของ range(start, stop) จะไม่ถูกรวมอยู่ในลำดับ ดังนั้น range(1, 5) จึงแสดงเฉพาะตัวเลข 1, 2, 3 และ 4

```python
# Snippet B: Fixed Incorrect Range

for i in range(1, 6):
    print(i)
```

```python
# Buggy Snippet C: Misplaced Break
# for char in "hello":
#     if char == 'l':
#         print("Found 'l', stopping!")
#         break
#     print(char) # This will print 'h', 'e', 'l' then break
```

Identify the Bug: คำสั่ง print(char) ถูกวางไว้หลังเงื่อนไข if และคำสั่ง break ทำให้โปรแกรมหยุดการทำงานก่อนที่จะแสดงตัวอักษร l

Explain Why: เมื่อโปรแกรมวนลูปมาถึงตัวอักษร l เงื่อนไข char == 'l' จะเป็นจริง และคำสั่ง break จะหยุดการทำงานของลูปทันที ส่งผลให้โปรแกรมแสดงเพียงตัวอักษร h และ e โดยไม่ได้แสดงตัวอักษร l

```python
# Snippet C: Fixed Misplaced Break

for char in "hello":
    print(char)
    if char == 'l':
        break
```
